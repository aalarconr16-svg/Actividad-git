1. Función print()
Permite mostrar información, mensajes o resultados en la pantalla.
Sintaxis: print(valor)


```python
print("Bienvenidos al curso de Python")
```

2.Función len()
Permite conocer la cantidad de elementos que tiene un objeto, como una palabra o una lista.


```python
palabra = "Veterinaria"
cantidad = len(palabra)
print(cantidad)
```

Función type()
Permite identificar el tipo de dato de una variable.

```python
edad = 20
print(type(edad))
```

Función int()
Convierte un valor compatible en un número entero.

```python
edad = "22"
numero = int(edad)
print(numero + 3)
```

5. Función float()
Convierte un número en un valor decimal.


```python
peso = 15
decimal = float(peso)
print(decimal)
```

6. Función str()
Convierte un número u otro valor en una cadena de texto.

```python
edad = 20
mensaje = "Mi edad es " + str(edad)
print(mensaje)
```

7. Función input()
Permite solicitar información al usuario mediante el teclado.

```python
nombre = input("Ingrese su nombre: ")
print("Hola", nombre)
```

8. Función range()
Genera una secuencia de números que se puede utilizar en ciclos.

```python
for numero in range(1, 6):
    print(numero)
```

9. Función abs()
Devuelve el valor absoluto de un número, sin considerar su signo.

```python
numero = -25
print(abs(numero))
```

10. Función round()
Permite redondear un número decimal a la cantidad de posiciones indicada.


```python
promedio = 4.56789
print(round(promedio, 2))
```
