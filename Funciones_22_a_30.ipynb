```python
frutas = ["manzana", "pera", "plátano"]
for indice, fruta in enumerate(frutas):
    print(indice, fruta)
```

```python
nombres = ["Ana", "Luis"]
edades = [25, 30]
combinados = list(zip(nombres, edades))
print(combinados)  # Salida: [('Ana', 25), ('Luis', 30)]
```

```python
condiciones = [False, False, True]
print(any(condiciones))  # Salida: True
```

```python
condiciones = [True, True, True]
print(all(condiciones))  # Salida: True
```

```python
print(chr(65))  # Salida: 'A'
```

```python
print(ord('A'))  # Salida: 65
```

```python
print(bin(5))  # Salida: '0b101'
```

```python
print(hex(255))  # Salida: '0xff'
```

```python
texto = "Hola"
print(isinstance(texto, str))  # Salida: True
```

```python
print(bool(0))    # Salida: False
print(bool(100))  # Salida: True
```
