## 11. Función max()

**Descripción:** La función `max()` devuelve el elemento de mayor valor de un conjunto de datos o de un iterable, como una lista.

**Sintaxis:**
`max(iterable)`

**Ejemplo:** Se utiliza para encontrar el número más grande de una lista.

```python
numeros = [4, 9, 2, 12, 1]

resultado = max(numeros)

print("El número mayor es:", resultado)
```

## 12. Función min()

**Descripción:** La función `min()` devuelve el elemento de menor valor de un conjunto de datos o de un iterable, como una lista.

**Sintaxis:**
`min(iterable)`

**Ejemplo:** Se utiliza para encontrar el número más pequeño de una lista.

```python
numeros = [4, 9, 2, 12, 1]

resultado = min(numeros)

print("El número menor es:", resultado)
```

## 13. Función sum()

**Descripción:** La función `sum()` permite sumar todos los elementos numéricos de un iterable, como una lista o una tupla.

**Sintaxis:**
`sum(iterable)`

**Ejemplo:** Se utiliza para calcular la suma total de varios números.

```python
valores = [10, 20, 30]

resultado = sum(valores)

print("La suma total es:", resultado)
```

## 14. Función sorted()

**Descripción:** La función `sorted()` devuelve una nueva lista con los elementos de un iterable ordenados. Por defecto, los organiza de menor a mayor.

**Sintaxis:**
`sorted(iterable)`

**Ejemplo:** Se utiliza para ordenar una lista de números de menor a mayor.

```python
numeros = [5, 2, 9, 1]

resultado = sorted(numeros)

print("Lista ordenada:", resultado)
```

## 15. Función list()

**Descripción:** La función `list()` permite crear una lista o convertir otro tipo de iterable, como una tupla, en una lista.

**Sintaxis:**
`list(iterable)`

**Ejemplo:** Se utiliza para convertir una tupla en una lista.

```python
numeros = (1, 2, 3)

resultado = list(numeros)

print("Lista convertida:", resultado)
```

## 16. Función dict()

**Descripción:** La función `dict()` permite crear un diccionario, una estructura que almacena información mediante pares de clave y valor.

**Sintaxis:**
`dict(clave=valor)`

**Ejemplo:** Se utiliza para crear un diccionario con información de una persona.

```python
persona = dict(nombre="Ana", edad=28)

print("Información de la persona:", persona)
```

## 17. Función set()

**Descripción:** La función `set()` permite crear un conjunto de elementos únicos. Si existen valores repetidos, los elimina automáticamente.

**Sintaxis:**
`set(iterable)`

**Ejemplo:** Se utiliza para eliminar los números duplicados de una lista.

```python
numeros = [1, 2, 2, 3, 3, 3]

resultado = set(numeros)

print("Conjunto sin duplicados:", resultado)
```

## 18. Función pow()

**Descripción:** La función `pow()` permite elevar un número a una potencia determinada. Es equivalente a utilizar el operador `**`.

**Sintaxis:**
`pow(base, exponente)`

**Ejemplo:** Se utiliza para calcular la potencia de un número.

```python
base = 2
exponente = 3

resultado = pow(base, exponente)

print("El resultado de la potencia es:", resultado)
```

## 19. Función help()

**Descripción:** La función `help()` permite consultar la documentación y obtener información sobre cómo utilizar funciones, módulos y otros elementos de Python.

**Sintaxis:**
`help(objeto)`

**Ejemplo:** Se utiliza para consultar la documentación de una función integrada de Python.

```python
help(print)
```

## 20. Función dir()

**Descripción:** La función `dir()` devuelve una lista con los nombres de los atributos y métodos disponibles de un objeto.

**Sintaxis:**
`dir(objeto)`

**Ejemplo:** Se utiliza para consultar los métodos y atributos disponibles de una cadena de texto.

```python
texto = "Hola"

resultado = dir(texto)

print("Métodos y atributos disponibles:", resultado)
```

## 21. Función enumerate()

**Descripción:** La función `enumerate()` permite recorrer un iterable y obtener al mismo tiempo el índice (posición) y el valor de cada elemento.

**Sintaxis:**
`enumerate(iterable)`

**Ejemplo:** Se utiliza para recorrer una lista de frutas mostrando su posición y su nombre.

```python
frutas = ["manzana", "pera", "plátano"]

for indice, fruta in enumerate(frutas):
    print(indice, fruta)
```
